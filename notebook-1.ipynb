```python
lines = [
    "ID,Nazwa,Stan\n",   # To będzie nasz header (pamiętaj o \n!)
    "1,Serwer_A,OK\n",
    "2,Serwer_B,Error\n",
    "3,Serwer_C,OK\n"
]
```

```python
with open('lines.txt','w') as f:
    f.writelines(lines)
    print('Utworzono plik "lines.txt"')
```

```python
lines = 'lines.txt'

with open(lines, 'r') as file:
    header = file.readline()
    print(f'Ignoruje naglowek: {header.strip()}')
    print('-'*20)
    for line in file:
        print(f'Przetwarzam: {line.strip()}')
```

```python
with open(lines,'r') as file:
    print('---Czytanie1----')
    print(file.read())

    print('---Czytanie2----')
    file.seek(0)
    fragment = file.read(5)
    print(f'Pierwsze 5 znakow: {fragment}')
    print(f'Pozycja kursora: {file.tell()}')
    
```

```python
with open(lines,'r+') as file:
    file.readline()
    reszta_danych = file.read()
    file.seek(0)

    nowy_header = 'ID,System,Status\n'
    file.write(nowy_header)
    file.write(reszta_danych)
    file.truncate()
    print('Zaktualizowano naglowek!')


    with open(lines,'r') as file:
        print('Tresc po edycji:\n'+file.read())
```
