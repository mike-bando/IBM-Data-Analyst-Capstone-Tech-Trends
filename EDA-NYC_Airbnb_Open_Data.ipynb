```python
from dash import Dash, html, dcc, callback, Output, Input
import plotly.express as px
import pandas as pd

# 1. Przygotowanie danych (wykorzystujemy Twoje przefiltrowane dane)
df = pd.read_csv('../Dane/AB_NYC_2019.csv')

# Inicjalizacja aplikacji Dash dla Jupytera
app = Dash(__name__)

# 2. Layout - Co widzimy na ekranie
app.layout = html.Div([
    html.H2("Analiza cen Airbnb w Nowym Jorku"),

    html.Label("Wybierz grupę dzielnic:"),
    dcc.Dropdown(
        options=df['neighbourhood_group'].unique(),
        value='Manhattan', # Wartość startowa
        id='dropdown-selection'
    ),

    dcc.Graph(id='graph-content')
])

# 3. Callback - Logika interakcji (Serce Dasha)
@callback(
    Output('graph-content', 'figure'),
    Input('dropdown-selection', 'value')
)
def update_graph(selected_group):
    # Filtrowanie danych na żywo
    filtered_df = df[df['neighbourhood_group'] == selected_group]

    # Tworzenie wykresu w Plotly
    fig = px.histogram(
        filtered_df,
        x='neighbourhood',
        y='price',
        histfunc='avg',
        title=f'Średnie ceny w: {selected_group}',
        labels={'price': 'Średnia Cena ($)', 'neighbourhood': 'Okolica'}
    ).update_xaxes(categoryorder='total descending')

    return fig

# 4. Uruchomienie wewnątrz notatnika
if __name__ == '__main__':
    app.run(jupyter_mode="inline") # To sprawi, że Dash pojawi się pod komórką
```
