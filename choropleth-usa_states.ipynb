```python
df.head()
```

```python
trace = go.Choropleth(
    colorscale = 'tealrose',
    locations = df['COUNTRY'],
    locationmode = 'country names',
    z = df['GDP (BILLIONS)'],
    text = df['COUNTRY'],
    marker = dict(line = dict(color = 'rgb(255,255,255)', width = 2)),
    colorbar = {'title':'World GDP'}
)
```

```python
layout = dict(title = '2014 World GDP',  width = 950,
    height = 600,
              geo = dict(showframe = False, 
                         projection = {'type':'natural earth'},     
                         showland = True,
                         landcolor = 'rgb(211, 211, 211)'))
```

```python
layout
```

```python
choromap2 = go.Figure(data=[trace], layout=layout)
```

```python
iplot(choromap2)
```
