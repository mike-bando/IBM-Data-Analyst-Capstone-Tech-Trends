#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.

Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df[['YearsCodePro','JobSat']]
```

```python
df['YearsCodePro'].value_counts().sort_index(ascending=False)
df['YearsCodePro'] = df['YearsCodePro'].replace('More than 50 years', 51).replace('Less than 1 year', 0)
df['YearsCodePro'] = df['YearsCodePro'].astype('float')
```

```python
def categorize_experience(years):
    if pd.isnull(years):
        return "Unknown"  # Missing values
    elif years <= 5:
        return "0-5 years (Junior)"
    elif years <= 10:
        return "6-10 years (Mid)"
    elif years <= 20:
        return "11-20 years (Senior)"
    else:
        return "20+ years (Veteran)"
```

```python
df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)
```

```python
# - Calculate the median `JobSat` for each range.

df[['JobSat','ExperienceLevel']]
```

```python
satisfaction_results = df.groupby('ExperienceLevel')['JobSat'].median()
print(satisfaction_results)
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(x='ExperienceLevel', y='JobSat', data=df)

plt.title('Job Satisfaction by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Job Satisfaction Score (0-10)')
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
sns.set_style("whitegrid")
plt.figure(figsize=(12,6))

sns.countplot(x='JobSat', data=df,palette="viridis", hue='JobSat')
plt.title('Distribution of Job Satisfaction Scores')
plt.ylabel("Number of Respondents")
plt.xlabel('Job Satisfaction Score (0-10)')
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
sns.set_style("whitegrid")
plt.figure(figsize=(12,6))

sns.countplot(x='RemoteWork', data=df)
plt.title('Distribution of Work Models')
plt.ylabel("Frequency")
plt.xlabel('Work Models')
plt.show()
```

```python
# Prosta metoda: przypisz "Full-time" jeśli ciąg zawiera to słowo, inaczej "Other"
df['Emp_Simple'] = df['Employment'].apply(lambda x: 'Full-time' if 'full-time' in str(x).lower() else 'Other/Part-time')
# Tabela krzyżowa: Modele pracy vs Uproszczone zatrudnienie
ct_employment = pd.crosstab(df['Emp_Simple'], df['RemoteWork'])

# Wyświetl tabelę w notebooku
print(ct_employment)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(ct_employment, annot=True, fmt="d", cmap="YlGnBu")
plt.title('Remote Work Preferences by Employment Status')
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# 1. Wybieramy Top 5 krajów
top_countries = df['Country'].value_counts().nlargest(5).index
df_languages = df[df['Country'].isin(top_countries)].copy()

# 2. Rozbijamy języki oddzielone średnikami
df_languages['Language'] = df_languages['LanguageHaveWorkedWith'].str.split(';')
df_languages = df_languages.explode('Language')

# 3. Wybieramy Top 10 języków
top_langs = df_languages['Language'].value_counts().nlargest(10).index

# --- TUTAJ BYŁ BŁĄD - Poprawka: najpierw filtrujemy całą tabelę ---
df_final_langs = df_languages[df_languages['Language'].isin(top_langs)]

# 4. Teraz tworzymy tabelę krzyżową z JUŻ PRZEFILTROWANEJ tabeli
ct_lang_region = pd.crosstab(df_final_langs['Language'], df_final_langs['Country'])

# 5. Wizualizacja
plt.figure(figsize=(12, 8))
sns.heatmap(ct_lang_region, annot=True, fmt="d", cmap="magma")
plt.title('Top 10 Programming Languages in Top 5 Countries')
plt.ylabel('Programming Language')
plt.xlabel('Country')
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
plt.figure(figsize=(10, 6))
# Regplot doda nam linię trendu, co ułatwi interpretację EDA
sns.regplot(x='YearsCodePro', y='JobSat', data=df,
            scatter_kws={'alpha':0.1}, line_kws={'color':'red'})

plt.title('Correlation between Years of Professional Coding and Job Satisfaction')
plt.xlabel('Years of Experience')
plt.ylabel('Satisfaction Score (0-10)')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
# Tabela krzyżowa wykształcenia i uproszczonego zatrudnienia
ct_ed_emp = pd.crosstab(df['EdLevel'], df['Emp_Simple'])

# Normalizujemy dane (procenty), żeby łatwiej było porównać grupy o różnej liczebności
ct_ed_emp_pct = ct_ed_emp.div(ct_ed_emp.sum(1), axis=0)

ct_ed_emp_pct.plot(kind='barh', stacked=True, figsize=(12, 8), color=['#4e79a7', '#f28e2b'])
plt.title('Employment Type by Educational Background (Percentage Share)')
plt.xlabel('Proportion')
plt.ylabel('Education Level')
plt.legend(title='Employment Status', bbox_to_anchor=(1.05, 1))
plt.show()
```
