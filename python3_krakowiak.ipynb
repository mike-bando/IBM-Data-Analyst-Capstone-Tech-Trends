```python
# Funkcja sumujaca liste liczb dodatnich
# ----------------------------------------
# Napisz funkcję, która jako argument przyjmuje listę liczb i zwraca sumę wszystkich liczb dodatnich na liście. 
# Funkcja powinna ignorować wszelkie liczby ujemne lub zera na liście.

def SumaListyLiczb(lista):
    suma = 0
    for x in lista:
        if x <= 0:
            continue
        else:
            suma += x
    print('Suma liczb z listy: ')
    return suma

lista = [0,-1,5,15,0,-5,1]
```

```python
# Program, który liczy sumę wszystkich liczb od 1 do podanej liczby
# np. dla 5
# 1+2+3+4+5
# zwróci 15

# Modul TIME
# perf_counter() -> float
import time
def sumuj_do(wybor):
    suma = 0
    # wybor = int(input('Podaj liczbe, do ktorej bedziemy sumowac wartosci (zaczynajac od 1): '))
    for liczba in range(1,wybor+1):
        suma += liczba
    return suma

def sumuj_do2(wybor2):
    # wybor2 = int(input("Podaj wartosc do zsumowania od 1: "))
    return sum([wybor2 for wybor2 in range(1,wybor2+1)])

def sumuj_do3(wybor3):
    suma3 = 0
    # wybor3 = int(input("Podaj wartosc do zsumowania od 1: "))
    suma3 = ((1+wybor3) / 2) * wybor3
    return suma3

def zakoncz_timer(start):
    end = time.perf_counter()
    return end-start

# funkcja jako argument funkcji
def function_performance(func, arg, how_many_times = 1): #how_many_times - domyslny argument, ktorego nie trzeba podawac przy wywolaniu, ale mozna go zmienic podajac nowa wartosc
    sum = 0
    for i in range(0, how_many_times):
        start = time.perf_counter()
        func(arg) #przesyla funkcje i argument do wywolania
        end = time.perf_counter()
        sum = sum + (end-start)
        # print(sum)
    return sum


print(function_performance(sumuj_do, 500000, 25))
# print(function_performance(sumuj_do2, 5000))
# print(function_performance(sumuj_do3, 5000))

```

```python
# czy dana liczba istnieje w liscie/zbiorze danych?

def znajdz_liczbe(x, seq):
    for liczba in seq:
        if x == liczba:
            return True
    return False
        
    
```

```python
# any - jakikolwiek
def any_even(lista):
    return any([x % 2 == 0 for x in lista])

print(any_even(numbers1))
print(any_even(numbers2))
```
